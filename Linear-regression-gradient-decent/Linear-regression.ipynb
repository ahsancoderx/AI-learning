```python
import numpy as np
```

```python
leanring_rate = 0.01 #step size
epoche = 5 #number of ittration
X = np.array([1,2,3,4,5]) #input data
Y = np.array([3,4,2,5,6]) # target data
```

```python
m,b = 0.0 , 0.0 # inital paramater
n = len(X) #number fo parmater

```

```python
for i in range(epoche):
    Y_pred = m*X + b
    cost = ((Y_pred - Y) ** 2).mean()
    derivative_m = (-2/n) * sum(X *(Y-Y_pred)) # DERIVATIVE WITH RESPECT TO m
    derivative_c = (-2/n) * sum(Y-Y_pred) # DERIVATIVE WITH RESPECT TO C
    m -= leanring_rate * derivative_m
    b -= leanring_rate * derivative_c
    print(f"Epoch {i+1}/{epoche}, cost : {cost} , m: {m}, b: {b}")
```

Topic 102 | Linear Regression - Implementation with Scikit-Learn

```python
from sklearn.linear_model import LinearRegression
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([[1],[3],[4],[5],[7]])
Y = np.array([3,5,7,9,11])
```

```python
# scatter plot before reqression
plt.figure(figsize=(10,6))
plt.scatter(X,Y, color ='blue')
plt.title("Scatter plot efore regression")
plt.xlabel("X values")
plt.ylabel("y values")
plt.show()
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X, Y)

Y_pred = model.predict(X)

print("predict value:", Y_pred)
```

```python
# scatter and best fit line
plt.figure(figsize=(6,4))
plt.scatter(X,Y , color='blue' , label="original data")
plt.plot(X, Y_pred, color="red" , label="best fit line")
plt.title("scatter plot with regression and best line")
plt.xlabel("x value")
plt.ylabel("y value")
plt.legend()
plt.show()
```

```python
m=model.coef_
b=model.intercept_
```
