```python

salary = 60000
# Check tax rate using conditional statements
if salary < 30000:
    tax_rate = 0.05
elif salary <= 70000:
    tax_rate = 0.15
else:
    tax_rate = 0.25
# Calculate tax and final salary
tax = salary * tax_rate
final_salary = salary - tax
# Display results
print("\n------ Salary Details ------")
print("Original Salary:", salary)
print("Tax Rate:", tax_rate * 100, "%")
print("Tax Amount:", tax)
print("Final Salary after Tax:", final_salary)
```

```python
class Message:
    def __init__(self, sender, text):
        self.sender = sender
        self.text = text

    def display(self):
        return f"{self.sender}: {self.text}"

class User:
    def __init__(self, name):
        self.name = name

    def send_message(self, chatroom, text):
        message = Message(self.name, text)
        chatroom.add_message(message)

class ChatRoom:
    def __init__(self):
        self.users = []
        self.messages = []

    # User joins the chat room
    def join(self, user):
        self.users.append(user)
        print(f"{user.name} joined the chat room.")

    # User leaves the chat room
    def leave(self, user):
        if user in self.users:
            self.users.remove(user)
            print(f"{user.name} left the chat room.")
        else:
            print(f"{user.name} is not in the chat room.")

    # Add message to chat history
    def add_message(self, message):
        self.messages.append(message)
        print(message.display())

    # Display all chat history
    def show_chat_history(self):
        print("\n----- Chat History -----")
        if len(self.messages) == 0:
            print("No messages available.")
        else:
            for message in self.messages:
                print(message.display())


# Create Chat Room
chatroom = ChatRoom()

# Create Users
user1 = User("Ahsan")
user2 = User("Ali")

# Users join the chat room
chatroom.join(user1)
chatroom.join(user2)

# Users send messages
user1.send_message(chatroom, "Hello everyone!")
user2.send_message(chatroom, "Hi Ahsan! How are you?")
user1.send_message(chatroom, "I am fine. Thank you!")

# Display chat history
chatroom.show_chat_history()

# User leaves the chat room
chatroom.leave(user2)
```
