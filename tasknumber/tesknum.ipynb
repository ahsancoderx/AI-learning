```python
import pandas as pd
df = pd.read_csv("HousePricePrediction.csv")
print(df.head(10))
```

```python
#Data Exploration
print("disply shape")
print(df.shape)
```

```python
#display data type
print("display data type")
print(df.dtypes)
```

```python
print("summerize data")
print(df.describe())
```

```python
#Data Cleaning
print("missiong value")
print(df.isnull().sum())
```

```python
#Handle Missing Values
print("handle missing value")
# Fill numerical missing values with mean
df.fillna(df.mean(numeric_only=True), inplace=True)
```

```python
print("Duplicate Rows:")
print(df.duplicated().sum())
```

```python
df.drop_duplicates(inplace=True)

print("Duplicates Removed")
```

```python
#Feature Selection
if 'Id' in df.columns:
    df.drop('Id', axis=1, inplace=True)

print(df.head())
```

```python
#Data Preprocessing
# Convert categorical variables into numerical form
df = pd.get_dummies(df, drop_first=True)
print(df.head())
```

```python
print("Final Prepared Dataset:")
print(df.head())
```

```python
#Save Final Dataset (Important)
df.to_csv("Prepared_HousePricePrediction.csv", index=False)

print("Prepared dataset saved successfully.")
```
